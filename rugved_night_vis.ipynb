```python
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
from tqdm import tqdm
import glob
# direct model from google
class enhance_net_nopool(nn.Module):
    def __init__(self):
        super(enhance_net_nopool, self).__init__()
        self.relu = nn.ReLU(inplace=True)
        n = 32
        self.e_conv1 = nn.Conv2d(3, n, 3, 1, 1)
        self.e_conv2 = nn.Conv2d(n, n, 3, 1, 1)
        self.e_conv3 = nn.Conv2d(n, n, 3, 1, 1)
        self.e_conv4 = nn.Conv2d(n, n, 3, 1, 1)
        self.e_conv5 = nn.Conv2d(n * 2, n, 3, 1, 1)
        self.e_conv6 = nn.Conv2d(n * 2, n, 3, 1, 1)
        self.e_conv7 = nn.Conv2d(n * 2, 24, 3, 1, 1)

    def forward(self, x):
        x1 = self.relu(self.e_conv1(x))
        x2 = self.relu(self.e_conv2(x1))
        x3 = self.relu(self.e_conv3(x2))
        x4 = self.relu(self.e_conv4(x3))
        x5 = self.relu(self.e_conv5(torch.cat([x3, x4], 1)))
        x6 = self.relu(self.e_conv6(torch.cat([x2, x5], 1)))
        x_r = torch.tanh(self.e_conv7(torch.cat([x1, x6], 1)))
        r1, r2, r3, r4, r5, r6, r7, r8 = torch.split(x_r, 3, dim=1)
        x = x + r1 * (x ** 2 - x)
        x = x + r2 * (x ** 2 - x)
        x = x + r3 * (x ** 2 - x)
        enhance_image_1 = x + r4 * (x ** 2 - x)
        x = enhance_image_1 + r5 * (enhance_image_1 ** 2 - enhance_image_1)
        x = x + r6 * (x ** 2 - x)
        x = x + r7 * (x ** 2 - x)
        enhance_image = x + r8 * (x ** 2 - x)
        return enhance_image
class LOLDataset(Dataset):
    def __init__(self, low_dir, high_dir):
        self.low_paths = sorted(glob.glob(os.path.join(low_dir, "*.png")))
        self.high_paths = sorted(glob.glob(os.path.join(high_dir, "*.png")))
        self.transform = transforms.Compose([
            transforms.Resize((256, 256)),
            transforms.ToTensor(),
        ])

    def __len__(self):
        return len(self.low_paths)
    def __getitem__(self, idx):
        low = Image.open(self.low_paths[idx]).convert("RGB")
        high = Image.open(self.high_paths[idx]).convert("RGB")
        return self.transform(low), self.transform(high)
def train():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = enhance_net_nopool().to(device)
    criterion = nn.L1Loss()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

    dataset = LOLDataset(
        low_dir=r"C:\Users\vedhr\datasets\our485\low",
        high_dir=r"C:\Users\vedhr\datasets\our485\high")
    loader = DataLoader(dataset, batch_size=4, shuffle=True)

    for epoch in range(10):
        model.train()
        epoch_loss = 0
        for low_img, high_img in tqdm(loader, desc=f"Epoch {epoch+1}"):
            low_img, high_img = low_img.to(device), high_img.to(device)
            enhanced = model(low_img)
            loss = criterion(enhanced, high_img)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()

        print(f"Epoch {epoch+1} Loss: {epoch_loss:.4f}")

    torch.save(model.state_dict(), "zero_dce_trained.pth")

if __name__ == "__main__":
    train()
```

```python
import os
import cv2
import time
import numpy as np
import torch
from ultralytics import YOLO
from torchreid.utils import FeatureExtractor
from scipy.spatial.distance import cosine
from model import enhance_net_nopool  # Zero-DCE model


# ==============================
# PARAMETERS
# ==============================
THRESHOLD = 0.30           # ReID cosine distance threshold
K = 3                      # Top-k averaging
MAX_EMBEDDINGS = 50        # Max live embeddings
EMBEDDING_DIR = r"C:\Users\vedhr\Downloads\re-id issue\embeddings"
BRIGHTNESS_THRESHOLD = 70  # Below this, Zero-DCE activates
ENHANCE_RES = (256, 256)   # Zero-DCE input resolution


# ==============================
# SETUP MODELS
# ==============================
device = torch.device("cuda")
print(f"🚀 Running on: {device}")

if not torch.cuda.is_available():
    print("⚠️  WARNING: CUDA not available, running on CPU")

print("📦 Loading YOLO and ReID models...")
# Force YOLO to use GPU
detector = YOLO("yolov8n.pt")
detector.to(device)

# Ensure ReID extractor uses GPU
extractor = FeatureExtractor(
    model_name='osnet_x1_0', 
    model_path='', 
    device=str(device)
)

# Load Zero-DCE model
enhancer = enhance_net_nopool().to(device)
enhancer.load_state_dict(torch.load("zero_dce_trained.pth", map_location=device))
enhancer.eval()
print("✅ Zero-DCE model loaded on", device)


# ==============================
# LOAD PRETRAINED EMBEDDINGS
# ==============================
pretrained_embeddings = []
expected_shape = (512,)
for fname in os.listdir(EMBEDDING_DIR):
    if fname.endswith(".npy"):
        emb = np.load(os.path.join(EMBEDDING_DIR, fname))
        if emb.shape == expected_shape:
            pretrained_embeddings.append(emb)

if not pretrained_embeddings:
    raise ValueError("❌ No valid pretrained embeddings found")

# Convert to GPU tensors for faster distance computation
pretrained_embeddings_tensor = torch.tensor(
    np.array(pretrained_embeddings), 
    dtype=torch.float32, 
    device=device
)
print(f"✅ Loaded {len(pretrained_embeddings)} pretrained embeddings on GPU.")


# ==============================
# HELPER FUNCTIONS
# ==============================
def get_brightness(frame):
    """Compute mean brightness."""
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    return np.mean(gray)


def enhance_frame(frame):
    """Run Zero-DCE enhancement on GPU."""
    h, w = frame.shape[:2]
    resized = cv2.resize(frame, ENHANCE_RES)
    tensor = torch.from_numpy(resized / 255.0).permute(2, 0, 1).unsqueeze(0).float().to(device)
    
    with torch.no_grad():
        enhanced = enhancer(tensor)[0]
    
    enhanced = enhanced.squeeze(0).permute(1, 2, 0).cpu().numpy()
    enhanced = np.clip(enhanced * 255, 0, 255).astype(np.uint8)
    return cv2.resize(enhanced, (w, h))


def get_embedding(frame, box):
    """Extract embedding on GPU."""
    x1, y1, x2, y2 = map(int, box)
    roi = frame[y1:y2, x1:x2]
    if roi.size == 0:
        return None
    
    roi = cv2.resize(roi, (128, 256))[:, :, ::-1]
    
    # Extractor returns tensor on GPU
    emb = extractor(roi)[0]
    
    # Keep as GPU tensor instead of converting to numpy
    return emb  # Returns torch tensor on GPU


def is_match_gpu(embedding, embeddings_tensor):
    """
    GPU-accelerated cosine distance computation.
    
    Args:
        embedding: torch.Tensor on GPU, shape (512,)
        embeddings_tensor: torch.Tensor on GPU, shape (N, 512)
    
    Returns:
        match: bool
        avg_dist: float
    """
    # Normalize vectors
    emb_norm = torch.nn.functional.normalize(embedding.unsqueeze(0), dim=1)
    refs_norm = torch.nn.functional.normalize(embeddings_tensor, dim=1)
    
    # Compute cosine similarity (1 - cosine distance)
    similarities = torch.mm(emb_norm, refs_norm.t()).squeeze(0)
    distances = 1 - similarities
    
    # Get top-K average
    top_k_dists, _ = torch.topk(distances, min(K, len(distances)), largest=False)
    avg_dist = top_k_dists.mean().item()
    
    return avg_dist < THRESHOLD, avg_dist


# ==============================
# CAPTURE SETUP
# ==============================

#cap = cv2.VideoCapture(0)
#url="rtsp://vedhr:1204@192.168.144.25:8554/main.264"
#cap=cv2.VideoCapture(url,cv2.CAP_FFMPEG)
cap=cv2.VideoCapture(0)
if not cap.isOpened():
    raise RuntimeError("❌ Failed to open webcam.")

# Set camera properties for better performance
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)
cap.set(cv2.CAP_PROP_FPS, 30)

boxes_in_frame = []
target_embeddings = []  # Store as GPU tensors
all_embeddings_list = pretrained_embeddings.copy()  # Keep numpy for backup
all_embeddings_tensor = pretrained_embeddings_tensor.clone()  # GPU tensors
selected_box = None


# ==============================
# STEP 1: COLLECT LIVE EMBEDDINGS
# ==============================
print("🧠 Phase 1: Collecting live embeddings of the target...")
count = 0

while count < MAX_EMBEDDINGS:
    ret, frame_raw = cap.read()
    if not ret:
        break

    # --- Lighting check ---
    brightness = get_brightness(frame_raw)
    if brightness < BRIGHTNESS_THRESHOLD:
        frame_proc = enhance_frame(frame_raw)
        cv2.putText(frame_proc, "Zero-DCE ON", (10, 40),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
    else:
        frame_proc = frame_raw.copy()

    boxes_in_frame.clear()
    
    # YOLO detection on GPU
    results = detector.predict(frame_proc, conf=0.3, verbose=False, imgsz=416, device=device)

    selected_box = None
    for det in results[0].boxes.data.cpu().numpy():
        x1, y1, x2, y2, _, cls = det
        if int(cls) != 0:
            continue
        
        box = (int(x1), int(y1), int(x2), int(y2))
        emb = get_embedding(frame_proc, box)
        if emb is None:
            continue
        
        # Use GPU-accelerated matching
        match, _ = is_match_gpu(emb, pretrained_embeddings_tensor)
        
        if match:
            selected_box = box
            target_embeddings.append(emb)  # Store GPU tensor
            
            # Update GPU tensor
            all_embeddings_tensor = torch.cat([
                all_embeddings_tensor, 
                emb.unsqueeze(0)
            ], dim=0)
            
            count += 1
            cv2.rectangle(frame_proc, (box[0], box[1]), (box[2], box[3]), (0, 255, 255), 2)
            cv2.putText(frame_proc, f"Collecting {count}/{MAX_EMBEDDINGS}", (10, 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
            break

    # --- Display both feeds ---
    side_by_side = np.hstack((cv2.resize(frame_raw, (480, 360)),
                              cv2.resize(frame_proc, (480, 360))))
    cv2.imshow("Camera (Raw) | Enhanced + Detection", side_by_side)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

print("✅ Live embeddings collected. Switching to tracking phase...")


# ==============================
# STEP 2: TRACKING
# ==============================
UPDATE_INTERVAL = 30
frame_count = 0

while True:
    ret, frame_raw = cap.read()
    if not ret:
        break

    brightness = get_brightness(frame_raw)
    if brightness < BRIGHTNESS_THRESHOLD:
        frame_proc = enhance_frame(frame_raw)
        cv2.putText(frame_proc, "Zero-DCE ON", (10, 40),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
    else:
        frame_proc = frame_raw.copy()

    boxes_in_frame.clear()
    
    # YOLO detection on GPU
    results = detector.predict(frame_proc, conf=0.3, verbose=False, imgsz=416, device=device)
    
    for det in results[0].boxes.data.cpu().numpy():
        x1, y1, x2, y2, _, cls = det
        if int(cls) != 0:
            continue
        box = (int(x1), int(y1), int(x2), int(y2))
        boxes_in_frame.append(box)
        cv2.rectangle(frame_proc, (box[0], box[1]), (box[2], box[3]), (255, 0, 0), 2)

    best_box = None
    best_score = 1.0
    
    for box in boxes_in_frame:
        emb = get_embedding(frame_proc, box)
        if emb is None:
            continue
        
        # GPU-accelerated matching
        match, score = is_match_gpu(emb, all_embeddings_tensor)
        
        if match and score < best_score:
            best_box = box
            best_score = score

    if best_box:
        cv2.rectangle(frame_proc, (best_box[0], best_box[1]),
                      (best_box[2], best_box[3]), (0, 255, 0), 2)
        cv2.putText(frame_proc, f"Target ({best_score:.2f})",
                    (best_box[0], best_box[1] - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

        frame_count += 1
        if frame_count % UPDATE_INTERVAL == 0:
            new_emb = get_embedding(frame_proc, best_box)
            if new_emb is not None:
                target_embeddings.append(new_emb)
                
                # Update GPU tensor
                all_embeddings_tensor = torch.cat([
                    all_embeddings_tensor,
                    new_emb.unsqueeze(0)
                ], dim=0)
                
                # Limit size
                if len(target_embeddings) > MAX_EMBEDDINGS:
                    target_embeddings.pop(0)
                
                if all_embeddings_tensor.shape[0] > 2 * MAX_EMBEDDINGS:
                    all_embeddings_tensor = all_embeddings_tensor[1:]

    # --- Show both streams ---
    side_by_side = np.hstack((cv2.resize(frame_raw, (480, 360)),
                              cv2.resize(frame_proc, (480, 360))))
    cv2.imshow("Camera (Raw) | Enhanced + Tracking", side_by_side)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

# Cleanup GPU memory
torch.cuda.empty_cache()
print("✅ Tracking completed. GPU memory cleared.")
```

```python
import os
import cv2
import time
import numpy as np
import torch
import onnxruntime as ort
from torchvision import transforms
from model import enhance_net_nopool
import matplotlib.pyplot as plt
import psutil
import GPUtil

# basic setup
start_time = time.time()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Running on:", device)

# load zero-dce model
enhancer = enhance_net_nopool().to(device)
enhancer.load_state_dict(torch.load("zero_dce_trained.pth", map_location=device))
enhancer.eval()

if not os.path.exists("yolov8n-face.onnx"):
    from ultralytics import YOLO
    YOLO("yolov8n-face.pt").export(format="onnx", dynamic=True, simplify=True)

# yolo onnx
providers = ["CUDAExecutionProvider"] if device.type == "cuda" else ["CPUExecutionProvider"]
yolo_sess = ort.InferenceSession("yolov8n-face.onnx", providers=providers)
yolo_input_name = yolo_sess.get_inputs()[0].name

# webcam
cap = cv2.VideoCapture(0)
print(f"Startup time: {(time.time() - start_time) * 1000:.2f} ms")

# tracking vars
transform = transforms.ToTensor()
prev_frame_time = time.time()
frame_count = 0
total_fps = 0
total_delay = 0
timestamps = []
fps_list = []
delay_list = []

def enhance_frame(frame):
    h, w = frame.shape[:2]
    frame_resized = cv2.resize(frame, (256, 256))
    norm = (frame_resized / 255.0).astype(np.float32)
    tensor = transform(norm).unsqueeze(0).to(device)

    with torch.no_grad():
        output = enhancer(tensor)[0]
    enhanced = output.squeeze(0).permute(1, 2, 0).cpu().numpy()
    enhanced = (enhanced * 255).clip(0, 255).astype(np.uint8)

    # quick sharpen + brightness
        # balanced enhancement (less harsh)
        # improved enhancement for cleaner night vision
        # clean + natural enhancement (no color tint)
    sharpen = np.array([[0, -0.3, 0],
                        [-0.3, 2.4, -0.3],
                        [0, -0.3, 0]])
    enhanced = cv2.filter2D(enhanced, -1, sharpen)

    # moderate brightness/contrast boost
    enhanced = cv2.convertScaleAbs(enhanced, alpha=1.2, beta=15)

    # smooth noise but keep edges
    enhanced = cv2.bilateralFilter(enhanced, 5, 40, 40)

    # optional: slight auto-adjust if lighting changes
    mean_brightness = np.mean(enhanced)
    if mean_brightness > 180:
        enhanced = cv2.convertScaleAbs(enhanced, alpha=0.9, beta=-10)
    elif mean_brightness < 70:
        enhanced = cv2.convertScaleAbs(enhanced, alpha=1.3, beta=20)



    return cv2.resize(enhanced, (w, h))

def run_yolo(img):
    h, w = img.shape[:2]
    img_resized = cv2.resize(img, (640, 640)).astype(np.float32) / 255.0
    input_blob = np.transpose(img_resized, (2, 0, 1))[None, ...]

    out = yolo_sess.run(None, {yolo_input_name: input_blob})[0]
    preds = np.squeeze(out).T

    confs = preds[:, 4]
    keep = confs > 0.075
    preds, confs = preds[keep], confs[keep]

    cx, cy, bw, bh = preds[:, 0], preds[:, 1], preds[:, 2], preds[:, 3]
    x1 = cx - bw / 2
    y1 = cy - bh / 2

    scale_x, scale_y = w / 640.0, h / 640.0
    boxes = [[int(x * scale_x), int(y * scale_y), int(bw_ * scale_x), int(bh_ * scale_y)]
             for x, y, bw_, bh_ in zip(x1, y1, bw, bh)]

    final = []
    if boxes:
        idxs = cv2.dnn.NMSBoxes(boxes, confs.tolist(), 0.075, 0.5)
        for i in idxs:
            x, y, bw_, bh_ = boxes[i]
            final.append((x, y, x + bw_, y + bh_, float(confs[i])))

    return final

# main loop
while True:
    ret, frame = cap.read()
    if not ret:
        break

    t0 = time.time()
    enhanced = enhance_frame(frame)
    boxes = run_yolo(enhanced)

    if boxes:
        x1, y1, x2, y2, conf = max(boxes, key=lambda b: b[4])
        cv2.rectangle(enhanced, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(enhanced, f"Face {conf:.2f}", (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    t1 = time.time()
    delay = (t1 - t0) * 1000
    fps = 1 / (t1 - prev_frame_time)
    prev_frame_time = t1

    frame_count += 1
    total_fps += fps
    total_delay += delay
    fps_list.append(fps)
    delay_list.append(delay)
    timestamps.append(t1 - start_time)
    
    cv2.putText(enhanced, f"FPS: {fps:.1f}", (10, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)
    cv2.putText(enhanced, f"Delay: {delay:.1f} ms", (10, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 0), 2)
    
    # show original but darkened to emphasize the night-vision effect
    orig_resized = cv2.resize(frame, (512, 512))
    
    # choose a darkness factor: 0.0 = fully black, 1.0 = original
    darkness_alpha = 0.40  # tweak between 0.25 - 0.5; lower -> darker
    overlay = np.zeros_like(orig_resized)
    darkened = cv2.addWeighted(orig_resized, 0.35, overlay, 0.65, 0)

    darkened = (orig_resized * darkness_alpha).astype(np.uint8)
    
    # optional: add a translucent label so it's clear it's the 'before' view
    cv2.putText(darkened, "Original ", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (200, 200, 200), 2, cv2.LINE_AA)
    
    dark_original = cv2.convertScaleAbs(darkened, alpha=0.8, beta=0)
    
    cv2.imshow("Original", dark_original)

    cv2.imshow("Enhanced + Detection", cv2.resize(enhanced, (512, 512)))


    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

# memory and GPU usage
mem = psutil.virtual_memory()
gpu = GPUtil.getGPUs()[0] if GPUtil.getGPUs() else None

# summary
print("\n- Performance Summary -")
import os, psutil


print(f"Total time: {time.time() - start_time:.2f} sec")
print(f"Frames: {frame_count}")
print(f"Avg FPS: {total_fps / frame_count:.2f}")
print(f"Avg Delay: {total_delay / frame_count:.2f} ms")
process = psutil.Process(os.getpid())
print(f"Script RAM Usage: {process.memory_info().rss / (1024 ** 2):.2f} MB")
print(f"GPU Load: {f'{gpu.load*100:.1f}%' if gpu else 'N/A'}")
print(f"GPU Mem: {f'{gpu.memoryUsed}MB / {gpu.memoryTotal}MB' if gpu else 'N/A'}")

# plotting
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(timestamps, fps_list, color='blue', label='FPS')
plt.xlabel("Time (s)")
plt.ylabel("FPS")
plt.title("FPS vs Time")
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(timestamps, delay_list, color='green', label='Delay (ms)')
plt.xlabel("Time (s)")
plt.ylabel("Delay")
plt.title("Delay vs Time")
plt.grid(True)
plt.legend()
plt.ylim(0, 30)
plt.yticks(np.arange(0, 31, 5))

plt.tight_layout()
plt.show()
```
